```python
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage
```

```python
data=load_iris()
X=data.data[:,[0,2]]
```

```python
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
Z=linkage(X_scaled, method='ward')
```

```python
M=linkage(X_scaled, method='single')
```

```python
N=linkage(X_scaled, method='complete')
```

```python
plt.figure(figsize=(10, 6))
dendrogram(Z)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points (Indices)")
plt.ylabel("Euclidean Distance")
plt.axhline(y=10,color='r',linestyle='--')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
dendrogram(M)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points (Indices)")
plt.ylabel("Euclidean Distance")
plt.axhline(y=10,color='r',linestyle='--')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
dendrogram(N)
plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Data Points (Indices)")
plt.ylabel("Euclidean Distance")
plt.axhline(y=10,color='r',linestyle='--')
plt.show()
```

```python
model1 = AgglomerativeClustering(n_clusters=3,linkage='ward')
labels1=model1.fit_predict(X_scaled)

model2 = AgglomerativeClustering(n_clusters=3,linkage='single')
labels2=model2.fit_predict(X_scaled)

model3 = AgglomerativeClustering(n_clusters=3,linkage='complete')
labels3=model3.fit_predict(X_scaled)
```

```python
plt.figure(figsize=(7,5))
plt.scatter(X_scaled[:,0],X_scaled[:,1],c=labels1,cmap='viridis',edgecolor='k')
plt.xlabel(data.feature_names[0])
plt.ylabel(data.feature_names[2])
plt.title("Agglomerative Clustering Results (K=3)")
plt.show()
```
